# **Calculation**

# **CAC - Customer Aquisition Cost**

```python
# tính CRM
```

```python
last_month = monthly_expenses.month.max()
```

```python
last_month_expenses = monthly_expenses[monthly_expenses['month'] == last_month]
```

```python
# cách 1
cond = last_month_expenses['item'] == 'Salesforce'
crm_cost = last_month_expenses[cond]['amount'].values[0]
print(crm_cost)
```

```python
# cách 2
CRM = monthly_expenses[(monthly_expenses['month'].dt.month == 3) & (monthly_expenses['item'] == 'Salesforce')]['amount'].sum()

print(CRM)
```

```python
# Marketing spending in mar 2023
Total_marketing_spending_mar = daily_marketing_spendings[daily_marketing_spendings['date'].dt.month == 3].spending.sum()
print(Total_marketing_spending_mar)
```

```python
# spending on each marketing chanel on mar
daily_marketing_spendings[(daily_marketing_spendings['date'].dt.month == 3)].groupby('channel')['spending'].sum()
```

```python
# total sales (  sales and marketing department in mar)

df_sales_marketing = payroll[payroll['department'].isin(['Sales', 'Marketing'])]
df_sales_marketing.head()
```

```python
total_sales_mar = df_sales_marketing[df_sales_marketing['month'].dt.month == 3].paid.sum()
print(total_sales_mar)
```

```python
df_sales_marketing[df_sales_marketing['month'].dt.month == 3].count()
# there are 6 staffs in sales and marketing department
```

```python
# total new customer in mar 2023
total_newcus_mar = receipts_history[receipts_history['date'].dt.month == 3].new_customer.sum()
print(total_newcus_mar)
```

```python
# CAC
CAC = (CRM + Total_marketing_spending_mar + total_sales_mar )/total_newcus_mar
print(CAC)
```

# **ARPU - Average revenue per user**

```python
# Total number of user in Mar
total_user_mar = receipts_history[receipts_history['date'].dt.month == 3].customer_id.count()
print(total_user_mar)
```

```python
# total revenue in mar
total_revenue_mar = receipts_history[receipts_history['date'].dt.month == 3].receipt_amount.sum()
print(total_revenue_mar)
```

```python
# ARPU
ARPU = total_revenue_mar / total_user_mar
print(ARPU)

 # meaning each customers generated $266.98 in revenue on average in mar
```

# **COGS - Cost of Goods Sold**

```python
# PRODUCTION COST  = AWS Hosting,  Google Cloud Storage, Atlassian Jira
```

```python
production_expense_tab = monthly_expenses[monthly_expenses['item'].isin(['AWS Hosting', 'Google Cloud Storage', 'Atlassian Jira'])]
```

```python
production_cost = production_expense_tab[
    production_expense_tab['month'].dt.month == 3
]['amount'].sum()
```

```python
production_cost
```

```python
# epense for each item (production cost) in mar
production_expense_tab[production_expense_tab['month'].dt.month == 3].groupby('item')['amount'].sum()
```

```python
from typing import SupportsFloat
# solfware cost : item slack and zoom
solfware_table = monthly_expenses[monthly_expenses['item'].isin(['Slack', 'Zoom'])]
solfware_cost = solfware_table[solfware_table['month'].dt.month == 3]['amount'].sum() * 0.6
solfware_cost
```

```python
# production_salary_cost = department - eninerring

production_salary_cost = payroll[(payroll['month'].dt.month == 3) & (payroll['department'] == 'Engineering')]['paid'].sum()
production_salary_cost
```

```python
payroll[(payroll['month'].dt.month == 3) & (payroll['department'] == 'Engineering')].count()
# there are 3 staff in Engineering deparment
```

```python
# COGS

COGS = production_cost + solfware_cost + production_salary_cost
print(COGS)
```

# **Gross Margin**

```python
# revenue
revenue = receipts_history[receipts_history['date'].dt.month == 3].receipt_amount.sum()
print(revenue)
```

```python
gross_margin = (revenue - COGS)/revenue * 100
print(gross_margin)
```

# **LTV - lifetime value**

```python
customer_lifespan = (customer_lifespan_data.churn_date - customer_lifespan_data.start_date).dt.days
```

```python
average_customer_lifespan = customer_lifespan.mean()
average_customer_lifespan_permonth = average_customer_lifespan/30
print(average_customer_lifespan_permonth)
```

```python
LTV = ARPU * average_customer_lifespan_permonth * gross_margin/100
print(LTV)
```
